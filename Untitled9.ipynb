```python
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
df.head()
```

```python
# We add errors='ignore' so if 'Over18' is already gone, the code won't crash.
cols_to_drop = ['Over18', 'EmployeeCount', 'StandardHours', 'EmployeeNumber']
df.drop(cols_to_drop, axis=1, inplace=True, errors='ignore')

print(f"\nDropped columns: {cols_to_drop}")
```

```python
# 3. Basic Cleaning
# Just run these standard commands to be safe.
df.drop_duplicates(inplace=True)  # Remove duplicates
df.dropna(inplace=True)           # Remove empty rows
```

```python
# PREPROCESSING STEP 2: ENCODING CATEGORICAL DATA
# Issue: Models cannot handle text strings.
# Solution: Label Encoding for Target, One-Hot for Features.
# ==========================================

# Encode Target Variable 'Attrition' (Yes -> 1, No -> 0)
label_encoder = LabelEncoder()
df['Attrition'] = label_encoder.fit_transform(df['Attrition'])
```

```python
# Define features (X) and target (y)
X = df.drop('Attrition', axis=1)
y = df['Attrition']
```

```python
# Apply One-Hot Encoding to categorical features (Department, JobRole, etc.)
# drop_first=True prevents multicollinearity
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
print("\n--- Data After Encoding (Text to Numbers) ---")
display(X_encoded.head())
```

```python
# PREPROCESSING STEP 3: FEATURE SCALING
# Issue: MonthlyIncome (20,000) dwarfs JobSatisfaction (4).
# Solution: Scale all numerical features to 0-1 range.
# ==========================================

scaler = MinMaxScaler()

# Fit the scaler to our features and transform them
X_scaled_array = scaler.fit_transform(X_encoded)
```

```python
# Convert back to a DataFrame for readability
X_final = pd.DataFrame(X_scaled_array, columns=X_encoded.columns)

print("\n--- Data After Scaling (0 to 1 Range) ---")
display(X_final.head())

print("\nPreprocessing Complete. Data shape:", X_final.shape)
```
